```python
import tensorflow as tf
from tensorflow.keras import layers, models

class SimpleConvNet(tf.keras.Model):
    def __init__(self):
        super(SimpleConvNet, self).__init__()
        self.conv = layers.Conv2D(
            filters=16,        # out_channels
            kernel_size=3,
            strides=1,
            padding='same',    # equivalent to PyTorch padding=1
            input_shape=(32, 32, 3)
        )
        self.relu = layers.ReLU()

    def call(self, x):
        x = self.conv(x)
        x = self.relu(x)
        return x
```

```python
def test_simple_convnet():
    model = SimpleConvNet()
    x = tf.random.normal((1, 32, 32, 3))  # batch=1, H=32, W=32, C=3
    y = model(x)
    assert y.shape == (1, 32, 32, 16), f"Unexpected output shape: {y.shape}"
    print("Test passed! Output shape:", y.shape)

test_simple_convnet()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

class ConvNetWithFC(tf.keras.Model):
    def __init__(self):
        super(ConvNetWithFC, self).__init__()
        # Conv + ReLU
        self.conv = layers.Conv2D(
            filters=16, kernel_size=3, strides=1, padding='same', input_shape=(32,32,3)
        )
        self.relu = layers.ReLU()
        # Flatten
        self.flatten = layers.Flatten()
        # Fully connected / Dense layer
        self.fc = layers.Dense(10)  # output 10 classes

    def call(self, x):
        x = self.conv(x)
        x = self.relu(x)
        x = self.flatten(x)
        x = self.fc(x)
        return x
```
